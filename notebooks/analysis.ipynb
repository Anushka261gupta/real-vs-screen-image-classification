## FFT Visualization

```python
def show_fft(image_path):
    img = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)

    f = np.fft.fft2(img)
    fshift = np.fft.fftshift(f)

    magnitude = 20 * np.log(np.abs(fshift) + 1)

    plt.figure(figsize=(10,4))

    plt.subplot(1,2,1)
    plt.imshow(img, cmap="gray")
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(magnitude, cmap="inferno")
    plt.title("FFT Magnitude")
    plt.axis("off")

    plt.show()
```

```python
show_fft(random.choice(real_images))
```

```python
show_fft(random.choice(screen_images))
```

```python
def edge_density(image_path):
    img = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)

    edges = cv2.Canny(img, 100, 200)

    density = np.sum(edges > 0) / edges.size

    return img, edges, density
```

```python
def show_edges(image_path):
    img, edges, density = edge_density(image_path)

    plt.figure(figsize=(10,4))

    plt.subplot(1,2,1)
    plt.imshow(img, cmap="gray")
    plt.title("Original")
    plt.axis("off")

    plt.subplot(1,2,2)
    plt.imshow(edges, cmap="gray")
    plt.title(f"Edges\nDensity={density:.4f}")
    plt.axis("off")

    plt.show()
```

```python
show_edges(random.choice(real_images))
show_edges(random.choice(screen_images))
```

```python
from sklearn.model_selection import train_test_split

all_images = real_images + screen_images
labels = [0] * len(real_images) + [1] * len(screen_images)

train_paths, val_paths, train_labels, val_labels = train_test_split(
    all_images,
    labels,
    test_size=0.2,
    random_state=42,
    stratify=labels
)

print("Train:", len(train_paths))
print("Validation:", len(val_paths))
print("Train Real:", train_labels.count(0))
print("Train Screen:", train_labels.count(1))
print("Val Real:", val_labels.count(0))
print("Val Screen:", val_labels.count(1))
```
